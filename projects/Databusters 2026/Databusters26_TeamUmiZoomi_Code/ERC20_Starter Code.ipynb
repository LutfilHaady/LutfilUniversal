### 📊 Understanding the Transaction Data Columns
Each row in the transaction dataset represents one ERC-20 token transfer event. The columns have the following meanings:
* `block_number` and `transaction_index`      
  Used to order transactions chronologically on the blockchain.
* `from_address`        
  The address that sends the token.
* `to_address`          
  The address that receives the token.
* `time_stamp`         
  The time of the transaction, recorded as a Unix timestamp (seconds since January 1, 1970, UTC).
* `contract_address`      
  Identifies which stablecoin or cryptocurrency is being transferred.
* `value`      
  The size of the transfer, representing the amount of capital moved.

## 1. Data Loading and Preparation

We'll load price data, transaction data, and event data to analyze the Terra-Luna collapse.

```python
# Define contract addresses for stablecoins
CONTRACT_ADDRESSES = {
    '0x6b175474e89094c44da98b954eedeac495271d0f': 'DAI',
    '0x8e870d67f660d95d5be530380d0ec0bd388289e1': 'PAX',
    '0xa0b86991c6218b36c1d19d4a2e9eb0ce3606eb48': 'USDC',
    '0xdac17f958d2ee523a2206206994597c13d831ec7': 'USDT',
    '0xa47c8bf37f92abed4a126bda807a7b7498661acd': 'USTC',
    '0xd2877702675e6ceb975b4a1dff9fb7baf4c91ea9': 'WLUNA'
}

# Key dates for Terra-Luna collapse (May 2022)
# UST depegged on May 9-10, 2022
COLLAPSE_START = pd.Timestamp('2022-05-09', tz='UTC')
COLLAPSE_END = pd.Timestamp('2022-06-01', tz='UTC')
PRE_COLLAPSE_START = pd.Timestamp('2022-04-01', tz='UTC')

print("Contract Addresses Mapping:")
for addr, name in CONTRACT_ADDRESSES.items():
    print(f"  {name}: {addr[:20]}...")
```

```python
# Load price data for all stablecoins
print("Loading price data...")
price_data = {}

with zipfile.ZipFile("ERC20-stablecoins.zip") as outer_zip:
    # Try different possible paths for the nested zip
    nested_zip_paths = ["price_data.zip", "ERC20-stablecoins/price_data.zip"]
    nested_zip_data = None
    for path in nested_zip_paths:
        try:
            nested_zip_data = outer_zip.read(path)
            break
        except KeyError:
            continue
    
    if nested_zip_data is None:
        print("  Error: Could not find price_data.zip in archive")
    else:
        with zipfile.ZipFile(io.BytesIO(nested_zip_data)) as inner_zip:
            for token in ['ustc', 'wluna', 'usdc', 'usdt', 'dai', 'pax']:
                try:
                    # Try different possible paths within the nested zip
                    paths_to_try = [f"price_data/{token}_price_data.csv", f"{token}_price_data.csv"]
                    df = None
                    for inner_path in paths_to_try:
                        try:
                            with inner_zip.open(inner_path) as f:
                                df = pd.read_csv(f)
                                break
                        except KeyError:
                            continue
                    
                    if df is not None:
                        df['timestamp'] = pd.to_datetime(df['timestamp'], unit='s', utc=True)
                        df['token'] = token.upper()
                        price_data[token.upper()] = df
                        print(f"  Loaded {token.upper()}: {len(df)} records")
                    else:
                        print(f"  Error: Could not find {token} price data")
                except Exception as e:
                    print(f"  Error loading {token}: {e}")

print(f"\nPrice data loaded for {len(price_data)} tokens")
```

```python
# Load event data (tweets/sentiment)
print("\nLoading event data...")
event_data = pd.DataFrame()
with zipfile.ZipFile("ERC20-stablecoins.zip") as z:
    # Try different possible paths
    for path in ["event_data.csv", "ERC20-stablecoins/event_data.csv"]:
        try:
            with z.open(path) as f:
                event_data = pd.read_csv(f, encoding='latin-1')
                event_data['timestamp'] = pd.to_datetime(event_data['timestamp'], unit='s', utc=True)
                print(f"  ✓ Loaded event data from {path}: {len(event_data)} events")
                print(f"  Event types: {event_data['type'].value_counts().to_dict()}")
                break
        except KeyError:
            continue
        except Exception as e:
            print(f"  ✗ Error loading {path}: {e}")
            continue
    if event_data.empty:
        print("  ⚠️ Event data not found in expected locations")
```

```python
# Load transaction data (we'll use chunks to handle large file)
print("\nLoading transaction data (this may take a while)...")
print("  Loading in chunks to manage memory...")

chunk_size = 100000
transactions = []
chunk_count = 0

with zipfile.ZipFile("ERC20-stablecoins.zip") as z:
    # Try the most recent version first - check both root and subdirectory
    for filename in ["token_transfers_V3.0.0.csv", 
                   "ERC20-stablecoins/token_transfers_V3.0.0.csv",
                   "token_transfers_V2.0.0.csv",
                   "ERC20-stablecoins/token_transfers_V2.0.0.csv",
                   "token_transfers.csv",
                   "ERC20-stablecoins/token_transfers.csv"]:
        try:
            with z.open(filename) as f:
                for chunk in pd.read_csv(f, chunksize=chunk_size):
                    chunk['time_stamp'] = pd.to_datetime(chunk['time_stamp'], unit='s', utc=True)
                    # Map contract addresses to token names
                    chunk['token'] = chunk['contract_address'].map(CONTRACT_ADDRESSES)
                    chunk = chunk[chunk['token'].notna()]  # Keep only known tokens
                    transactions.append(chunk)
                    chunk_count += 1
                    if chunk_count % 10 == 0:
                        print(f"    Processed {chunk_count} chunks...")
                print(f"  ✓ Loaded transactions from {filename}")
                break
        except KeyError:
            continue
        except Exception as e:
            print(f"  ✗ Error loading {filename}: {e}")
            continue

if transactions:
    tx_df = pd.concat(transactions, ignore_index=True)
    print(f"\n  Total transactions loaded: {len(tx_df):,}")
    print(f"  Date range: {tx_df['time_stamp'].min()} to {tx_df['time_stamp'].max()}")
    print(f"  Tokens: {tx_df['token'].value_counts().to_dict()}")
else:
    print("  ✗ No transaction data loaded")
    tx_df = pd.DataFrame()
```

## Question 1: When the Peg Breaks - Onset and Spread

We analyze:
1. **Onset**: When confidence began to break down (price depegging)
2. **Spread**: How panic propagated across participants (transaction flows, volumes)

```python
# ============================================================================
# QUESTION 1.1: IDENTIFYING THE ONSET - PRICE DEPEGGING
# ============================================================================

if 'USTC' in price_data:
    ustc_price = price_data['USTC'].copy()
    ustc_price = ustc_price.sort_values('timestamp')
    
    # Calculate deviation from $1 peg
    ustc_price['peg_deviation'] = (ustc_price['close'] - 1.0) * 100  # Percentage deviation
    ustc_price['peg_deviation_abs'] = abs(ustc_price['peg_deviation'])
    
    # Identify depegging threshold (e.g., > 1% deviation)
    depeg_threshold = 1.0  # 1% deviation
    ustc_price['is_depegged'] = ustc_price['peg_deviation_abs'] > depeg_threshold
    
    # Find first significant depegging event
    first_depeg = ustc_price[ustc_price['is_depegged']].iloc[0] if ustc_price['is_depegged'].any() else None
    
    print("=" * 80)
    print("QUESTION 1.1: ONSET OF THE CRISIS")
    print("=" * 80)
    
    if first_depeg is not None:
        print(f"\n🔴 First significant depegging detected:")
        print(f"   Date: {first_depeg['timestamp']}")
        print(f"   Price: ${first_depeg['close']:.4f}")
        print(f"   Deviation from peg: {first_depeg['peg_deviation']:.2f}%")
    
    # Find when price dropped below 0.95 (5% depeg)
    major_depeg = ustc_price[ustc_price['close'] < 0.95]
    if len(major_depeg) > 0:
        major_depeg_date = major_depeg.iloc[0]['timestamp']
        major_depeg_price = major_depeg.iloc[0]['close']
        print(f"\n🔴 Major depegging (< $0.95):")
        print(f"   Date: {major_depeg_date}")
        print(f"   Price: ${major_depeg_price:.4f}")
    
    # Plot price evolution
    fig, axes = plt.subplots(2, 1, figsize=(14, 10))
    
    # Plot 1: USTC Price
    ax1 = axes[0]
    ax1.plot(ustc_price['timestamp'], ustc_price['close'], 
             linewidth=2, label='USTC Price', color='red')
    ax1.axhline(y=1.0, color='green', linestyle='--', linewidth=2, label='$1.00 Peg')
    ax1.axhline(y=0.95, color='orange', linestyle='--', linewidth=1, alpha=0.7, label='5% Depeg Threshold')
    if first_depeg is not None:
        ax1.axvline(x=first_depeg['timestamp'], color='red', linestyle=':', 
                   linewidth=2, alpha=0.7, label='First Depeg')
    ax1.set_xlabel('Date', fontsize=12)
    ax1.set_ylabel('Price (USD)', fontsize=12)
    ax1.set_title('USTC Price Evolution: Onset of the Crisis', fontsize=14, fontweight='bold')
    ax1.legend(loc='best')
    ax1.grid(True, alpha=0.3)
    ax1.set_xlim([PRE_COLLAPSE_START, COLLAPSE_END])
    
    # Plot 2: Deviation from Peg
    ax2 = axes[1]
    ax2.plot(ustc_price['timestamp'], ustc_price['peg_deviation'], 
             linewidth=2, label='Deviation from $1.00 Peg', color='darkred')
    ax2.axhline(y=0, color='green', linestyle='--', linewidth=1, alpha=0.5)
    ax2.axhline(y=-1, color='orange', linestyle='--', linewidth=1, alpha=0.7)
    ax2.fill_between(ustc_price['timestamp'], -1, 1, alpha=0.1, color='green', label='Stable Zone (±1%)')
    if first_depeg is not None:
        ax2.axvline(x=first_depeg['timestamp'], color='red', linestyle=':', 
                   linewidth=2, alpha=0.7)
    ax2.set_xlabel('Date', fontsize=12)
    ax2.set_ylabel('Deviation from Peg (%)', fontsize=12)
    ax2.set_title('USTC Peg Deviation Over Time', fontsize=14, fontweight='bold')
    ax2.legend(loc='best')
    ax2.grid(True, alpha=0.3)
    ax2.set_xlim([PRE_COLLAPSE_START, COLLAPSE_END])
    
    plt.tight_layout()
    plt.savefig('q1_onset_price_depegging.png', dpi=300, bbox_inches='tight')
    plt.show()
    
    print(f"\n📊 Visualization saved as 'q1_onset_price_depegging.png'")
else:
    print("⚠️ USTC price data not available")
```

```python
# ============================================================================
# QUESTION 1.2: STRESS BECOMES VISIBLE - TRANSACTION VOLUMES AND FLOWS
# ============================================================================

if not tx_df.empty:
    # Filter to relevant time period
    tx_analysis = tx_df[
        (tx_df['time_stamp'] >= PRE_COLLAPSE_START) & 
        (tx_df['time_stamp'] <= COLLAPSE_END)
    ].copy()
    
    # Aggregate by day and token
    tx_analysis['date'] = tx_analysis['time_stamp'].dt.date
    daily_volumes = tx_analysis.groupby(['date', 'token'])['value'].agg(['sum', 'count']).reset_index()
    daily_volumes.columns = ['date', 'token', 'total_volume', 'transaction_count']
    daily_volumes['date'] = pd.to_datetime(daily_volumes['date'])
    
    print("\n" + "=" * 80)
    print("QUESTION 1.2: STRESS BECOMES VISIBLE - TRANSACTION PATTERNS")
    print("=" * 80)
    
    # Calculate USTC outflows (selling pressure)
    ustc_daily = daily_volumes[daily_volumes['token'] == 'USTC'].copy()
    if len(ustc_daily) > 0:
        # Find peak outflow day
        peak_outflow = ustc_daily.loc[ustc_daily['total_volume'].idxmax()]
        print(f"\n📈 Peak USTC transaction volume:")
        print(f"   Date: {peak_outflow['date'].date()}")
        print(f"   Volume: ${peak_outflow['total_volume']:,.2f}")
        print(f"   Transaction count: {peak_outflow['transaction_count']:,}")
    
    # Calculate flows to safe havens (USDC, USDT, DAI)
    safe_havens = ['USDC', 'USDT', 'DAI']
    safe_haven_flows = daily_volumes[daily_volumes['token'].isin(safe_havens)].copy()
    
    # Plot transaction volumes
    fig, axes = plt.subplots(3, 1, figsize=(14, 12))
    
    # Plot 1: USTC Volume
    ax1 = axes[0]
    if len(ustc_daily) > 0:
        ax1.plot(ustc_daily['date'], ustc_daily['total_volume'] / 1e6, 
                marker='o', linewidth=2, markersize=4, label='USTC Volume', color='red')
        ax1.set_ylabel('Volume (Millions USD)', fontsize=11)
        ax1.set_title('USTC Transaction Volume: Selling Pressure Indicator', fontsize=13, fontweight='bold')
        ax1.legend()
        ax1.grid(True, alpha=0.3)
        ax1.axvline(x=COLLAPSE_START, color='red', linestyle='--', alpha=0.5, label='Crisis Start')
    
    # Plot 2: Safe Haven Flows
    ax2 = axes[1]
    for token in safe_havens:
        token_data = safe_haven_flows[safe_haven_flows['token'] == token]
        if len(token_data) > 0:
            ax2.plot(token_data['date'], token_data['total_volume'] / 1e6, 
                    marker='o', linewidth=2, markersize=3, label=token, alpha=0.8)
    ax2.set_ylabel('Volume (Millions USD)', fontsize=11)
    ax2.set_title('Flight to Safety: Flows to USDC, USDT, DAI', fontsize=13, fontweight='bold')
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    ax2.axvline(x=COLLAPSE_START, color='red', linestyle='--', alpha=0.5)
    
    # Plot 3: Transaction Counts (Activity Level)
    ax3 = axes[2]
    for token in ['USTC'] + safe_havens:
        token_data = daily_volumes[daily_volumes['token'] == token]
        if len(token_data) > 0:
            ax3.plot(token_data['date'], token_data['transaction_count'], 
                    marker='o', linewidth=1.5, markersize=3, label=token, alpha=0.7)
    ax3.set_xlabel('Date', fontsize=12)
    ax3.set_ylabel('Transaction Count', fontsize=11)
    ax3.set_title('Transaction Activity: Panic Propagation Indicator', fontsize=13, fontweight='bold')
    ax3.legend()
    ax3.grid(True, alpha=0.3)
    ax3.axvline(x=COLLAPSE_START, color='red', linestyle='--', alpha=0.5)
    ax3.set_yscale('log')  # Log scale for better visualization
    
    plt.tight_layout()
    plt.savefig('q1_stress_transaction_patterns.png', dpi=300, bbox_inches='tight')
    plt.show()
    
    print(f"\n📊 Visualization saved as 'q1_stress_transaction_patterns.png'")
else:
    print("⚠️ Transaction data not available")
```

```python
# ============================================================================
# QUESTION 1.3: PANIC PROPAGATION - NET FLOWS BETWEEN TOKENS
# ============================================================================

if not tx_df.empty:
    # Analyze net flows: USTC outflows vs inflows to other tokens
    tx_analysis = tx_df[
        (tx_df['time_stamp'] >= PRE_COLLAPSE_START) & 
        (tx_df['time_stamp'] <= COLLAPSE_END)
    ].copy()
    
    # Calculate daily net flows for each token
    # Net flow = total received - total sent
    daily_flows = []
    
    for date in pd.date_range(start=PRE_COLLAPSE_START, end=COLLAPSE_END, freq='D'):
        date_str = date.date()
        day_data = tx_analysis[tx_analysis['time_stamp'].dt.date == date_str]
        
        for token in ['USTC', 'USDC', 'USDT', 'DAI', 'WLUNA']:
            token_data = day_data[day_data['token'] == token]
            if len(token_data) > 0:
                total_sent = token_data['value'].sum()  # All outflows
                # For simplicity, we use total volume as proxy for activity
                # In reality, we'd need to track specific address balances
                daily_flows.append({
                    'date': date,
                    'token': token,
                    'total_volume': token_data['value'].sum(),
                    'tx_count': len(token_data)
                })
    
    flows_df = pd.DataFrame(daily_flows)
    
    # Calculate relative changes (momentum)
    flows_df['volume_change'] = flows_df.groupby('token')['total_volume'].pct_change()
    flows_df['tx_change'] = flows_df.groupby('token')['tx_count'].pct_change()
    
    print("\n" + "=" * 80)
    print("QUESTION 1.3: PANIC PROPAGATION - FLOW DYNAMICS")
    print("=" * 80)
    
    # Identify propagation pattern
    if len(flows_df) > 0:
        # Find when USTC activity spiked
        ustc_flows = flows_df[flows_df['token'] == 'USTC']
        if len(ustc_flows) > 0:
            peak_ustc = ustc_flows.loc[ustc_flows['total_volume'].idxmax()]
            print(f"\n📊 Peak USTC activity:")
            print(f"   Date: {peak_ustc['date'].date()}")
            print(f"   Volume: ${peak_ustc['total_volume']:,.2f}")
        
        # Find when safe havens saw increased activity
        for token in ['USDC', 'USDT']:
            token_flows = flows_df[flows_df['token'] == token]
            if len(token_flows) > 0:
                peak_token = token_flows.loc[token_flows['total_volume'].idxmax()]
                print(f"\n📊 Peak {token} activity:")
                print(f"   Date: {peak_token['date'].date()}")
                print(f"   Volume: ${peak_token['total_volume']:,.2f}")
    
    # Visualization: Heatmap of activity by token and date
    if len(flows_df) > 0:
        pivot_volume = flows_df.pivot_table(
            index='date', 
            columns='token', 
            values='total_volume', 
            aggfunc='sum'
        ).fillna(0)
        
        # Normalize for better visualization
        pivot_volume_norm = pivot_volume.div(pivot_volume.max(axis=0), axis=1)
        
        fig, axes = plt.subplots(2, 1, figsize=(14, 10))
        
        # Heatmap 1: Absolute volumes
        ax1 = axes[0]
        sns.heatmap(pivot_volume.T / 1e6, ax=ax1, cmap='YlOrRd', 
                   cbar_kws={'label': 'Volume (Millions USD)'}, 
                   linewidths=0.5, linecolor='gray')
        ax1.set_title('Daily Transaction Volumes by Token: Panic Propagation', 
                     fontsize=14, fontweight='bold', pad=20)
        ax1.set_xlabel('Date', fontsize=12)
        ax1.set_ylabel('Token', fontsize=12)
        
        # Heatmap 2: Normalized (relative activity)
        ax2 = axes[1]
        sns.heatmap(pivot_volume_norm.T, ax=ax2, cmap='viridis', 
                   cbar_kws={'label': 'Normalized Activity (0-1)'}, 
                   linewidths=0.5, linecolor='gray')
        ax2.set_title('Normalized Activity Levels: Relative Panic Intensity', 
                     fontsize=14, fontweight='bold', pad=20)
        ax2.set_xlabel('Date', fontsize=12)
        ax2.set_ylabel('Token', fontsize=12)
        
        plt.tight_layout()
        plt.savefig('q1_panic_propagation_flows.png', dpi=300, bbox_inches='tight')
        plt.show()
        
        print(f"\n📊 Visualization saved as 'q1_panic_propagation_flows.png'")
else:
    print("⚠️ Transaction data not available for flow analysis")
```

## Question 3: Who Bears the Losses?

We analyze:
1. **Loss Identification**: Which addresses/participants suffered losses
2. **System Design Failures**: Why the system couldn't protect them
3. **Comparative Analysis**: Differences in design and outcomes

```python
# ============================================================================
# QUESTION 3.1: IDENTIFYING LOSSES - ADDRESS-LEVEL ANALYSIS
# ============================================================================

if not tx_df.empty:
    # Focus on USTC transactions during collapse
    collapse_period = tx_df[
        (tx_df['time_stamp'] >= COLLAPSE_START) & 
        (tx_df['time_stamp'] <= COLLAPSE_END) &
        (tx_df['token'] == 'USTC')
    ].copy()
    
    print("=" * 80)
    print("QUESTION 3.1: IDENTIFYING WHO BEARS THE LOSSES")
    print("=" * 80)
    
    if len(collapse_period) > 0:
        # Calculate net positions by address
        # Positive = net receiver (may have losses if price dropped)
        # Negative = net sender (may have losses if sold at low price)
        
        # Calculate inflows and outflows per address
        inflows = collapse_period.groupby('to_address')['value'].sum().reset_index()
        inflows.columns = ['address', 'total_received']
        
        outflows = collapse_period.groupby('from_address')['value'].sum().reset_index()
        outflows.columns = ['address', 'total_sent']
        
        # Merge to get net position
        address_flows = inflows.merge(outflows, on='address', how='outer').fillna(0)
        address_flows['net_position'] = address_flows['total_received'] - address_flows['total_sent']
        address_flows['total_activity'] = address_flows['total_received'] + address_flows['total_sent']
        
        # Identify large holders (potential loss bearers)
        # Large net receivers = holders who accumulated USTC during collapse
        large_holders = address_flows[address_flows['net_position'] > 10000].sort_values(
            'net_position', ascending=False
        )
        
        # Large net senders = those who sold during collapse (may have sold at loss)
        large_sellers = address_flows[address_flows['total_sent'] > 10000].sort_values(
            'total_sent', ascending=False
        )
        
        print(f"\n📊 Analysis of {len(collapse_period):,} USTC transactions during collapse")
        print(f"   Unique addresses: {len(address_flows):,}")
        
        print(f"\n🔴 Large Net Receivers (Accumulated USTC - Potential Loss Bearers):")
        print(f"   Top 10 addresses by net position:")
        for idx, row in large_holders.head(10).iterrows():
            print(f"   {row['address'][:20]}... | Net: ${row['net_position']:,.2f} | "
                  f"Total Activity: ${row['total_activity']:,.2f}")
        
        print(f"\n🔴 Large Sellers (Sold During Collapse - Potential Loss Bearers):")
        print(f"   Top 10 addresses by volume sold:")
        for idx, row in large_sellers.head(10).iterrows():
            print(f"   {row['address'][:20]}... | Sold: ${row['total_sent']:,.2f} | "
                  f"Net Position: ${row['net_position']:,.2f}")
        
        # Calculate aggregate statistics
        total_ustc_volume = collapse_period['value'].sum()
        print(f"\n📈 Aggregate Statistics:")
        print(f"   Total USTC volume during collapse: ${total_ustc_volume:,.2f}")
        print(f"   Average transaction size: ${collapse_period['value'].mean():,.2f}")
        print(f"   Median transaction size: ${collapse_period['value'].median():,.2f}")
        
        # Distribution analysis
        fig, axes = plt.subplots(2, 2, figsize=(16, 12))
        
        # Plot 1: Distribution of net positions
        ax1 = axes[0, 0]
        # Filter to reasonable range for visualization
        net_pos_vis = address_flows[
            (address_flows['net_position'] > -100000) & 
            (address_flows['net_position'] < 100000)
        ]['net_position']
        ax1.hist(net_pos_vis, bins=50, edgecolor='black', alpha=0.7, color='red')
        ax1.set_xlabel('Net Position (USD)', fontsize=11)
        ax1.set_ylabel('Number of Addresses', fontsize=11)
        ax1.set_title('Distribution of Net USTC Positions', fontsize=12, fontweight='bold')
        ax1.axvline(x=0, color='black', linestyle='--', linewidth=2)
        ax1.grid(True, alpha=0.3)
        
        # Plot 2: Top holders
        ax2 = axes[0, 1]
        top_holders_plot = large_holders.head(20)
        ax2.barh(range(len(top_holders_plot)), 
                top_holders_plot['net_position'] / 1e6,
                color='darkred', alpha=0.7)
        ax2.set_yticks(range(len(top_holders_plot)))
        ax2.set_yticklabels([addr[:15] + '...' for addr in top_holders_plot['address']], 
                           fontsize=8)
        ax2.set_xlabel('Net Position (Millions USD)', fontsize=11)
        ax2.set_title('Top 20 Net USTC Receivers (Loss Bearers)', fontsize=12, fontweight='bold')
        ax2.grid(True, alpha=0.3, axis='x')
        
        # Plot 3: Transaction size distribution
        ax3 = axes[1, 0]
        tx_sizes = collapse_period['value']
        # Log scale for better visualization
        ax3.hist(tx_sizes[tx_sizes > 0], bins=100, edgecolor='black', alpha=0.7, color='orange')
        ax3.set_xlabel('Transaction Size (USD)', fontsize=11)
        ax3.set_ylabel('Frequency', fontsize=11)
        ax3.set_title('Distribution of Transaction Sizes', fontsize=12, fontweight='bold')
        ax3.set_xscale('log')
        ax3.grid(True, alpha=0.3)
        
        # Plot 4: Daily loss exposure (volume held)
        ax4 = axes[1, 1]
        daily_ustc_volume = collapse_period.groupby(collapse_period['time_stamp'].dt.date)['value'].sum()
        daily_ustc_volume.index = pd.to_datetime(daily_ustc_volume.index)
        ax4.plot(daily_ustc_volume.index, daily_ustc_volume.values / 1e6, 
                marker='o', linewidth=2, markersize=6, color='red')
        ax4.set_xlabel('Date', fontsize=11)
        ax4.set_ylabel('Daily USTC Volume (Millions USD)', fontsize=11)
        ax4.set_title('Daily USTC Transaction Volume: Loss Exposure Indicator', 
                     fontsize=12, fontweight='bold')
        ax4.grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.savefig('q3_loss_identification.png', dpi=300, bbox_inches='tight')
        plt.show()
        
        print(f"\n📊 Visualization saved as 'q3_loss_identification.png'")
    else:
        print("⚠️ No USTC transactions found in collapse period")
else:
    print("⚠️ Transaction data not available")
```

```python
# ============================================================================
# QUESTION 3.2: ESTIMATING LOSSES - PRICE-BASED CALCULATION
# ============================================================================

if 'USTC' in price_data and not tx_df.empty:
    # Get price data during collapse
    collapse_prices = price_data['USTC'][
        (price_data['USTC']['timestamp'] >= COLLAPSE_START) & 
        (price_data['USTC']['timestamp'] <= COLLAPSE_END)
    ].copy()
    
    if len(collapse_prices) > 0:
        # Calculate price decline
        initial_price = collapse_prices.iloc[0]['close']
        final_price = collapse_prices.iloc[-1]['close']
        price_decline = (initial_price - final_price) / initial_price * 100
        
        print("\n" + "=" * 80)
        print("QUESTION 3.2: ESTIMATING LOSSES")
        print("=" * 80)
        
        print(f"\n💰 Price Decline Analysis:")
        print(f"   Initial price (May 9): ${initial_price:.4f}")
        print(f"   Final price (end of period): ${final_price:.4f}")
        print(f"   Decline: {price_decline:.2f}%")
        
        # Estimate losses for holders
        collapse_tx = tx_df[
            (tx_df['time_stamp'] >= COLLAPSE_START) & 
            (tx_df['time_stamp'] <= COLLAPSE_END) &
            (tx_df['token'] == 'USTC')
        ].copy()
        
        if len(collapse_tx) > 0:
            # Merge with price data to estimate losses
            collapse_tx['date'] = collapse_tx['time_stamp'].dt.date
            collapse_prices['date'] = collapse_prices['timestamp'].dt.date
            
            # Get price for each transaction day
            daily_prices = collapse_prices.groupby('date')['close'].mean().reset_index()
            daily_prices.columns = ['date', 'price']
            
            collapse_tx = collapse_tx.merge(daily_prices, on='date', how='left')
            collapse_tx['price'] = collapse_tx['price'].ffill()
            collapse_tx['price'] = collapse_tx['price'].fillna(initial_price)
            
            # Estimate losses: if someone received USTC at price X, and price dropped to Y
            # Loss = (X - Y) * amount
            # For simplicity, assume holders at start of collapse
            total_ustc_held = collapse_tx.groupby('to_address')['value'].sum()
            
            # Estimate loss per holder (assuming they held until end)
            estimated_losses = total_ustc_held * (initial_price - final_price)
            estimated_losses = estimated_losses[estimated_losses > 0].sort_values(ascending=False)
            
            print(f"\n💸 Estimated Losses (assuming holding from start to end):")
            print(f"   Total estimated losses: ${estimated_losses.sum():,.2f}")
            print(f"   Number of addresses with estimated losses: {len(estimated_losses):,}")
            print(f"   Average loss per affected address: ${estimated_losses.mean():,.2f}")
            print(f"   Median loss per affected address: ${estimated_losses.median():,.2f}")
            
            print(f"\n🔴 Top 20 Estimated Loss Bearers:")
            for idx, (address, loss) in enumerate(estimated_losses.head(20).items(), 1):
                print(f"   {idx:2d}. {address[:20]}... | Estimated Loss: ${loss:,.2f}")
            
            # Visualization
            fig, axes = plt.subplots(2, 2, figsize=(16, 12))
            
            # Plot 1: Price decline with volume overlay
            ax1 = axes[0, 0]
            ax1_twin = ax1.twinx()
            
            ax1.plot(collapse_prices['timestamp'], collapse_prices['close'], 
                    linewidth=3, color='red', label='USTC Price', alpha=0.8)
            ax1.axhline(y=1.0, color='green', linestyle='--', linewidth=2, 
                       alpha=0.5, label='$1.00 Peg')
            ax1.set_xlabel('Date', fontsize=11)
            ax1.set_ylabel('Price (USD)', fontsize=11, color='red')
            ax1.tick_params(axis='y', labelcolor='red')
            ax1.set_title('USTC Price Collapse: Loss Magnitude', fontsize=12, fontweight='bold')
            ax1.legend(loc='upper left')
            ax1.grid(True, alpha=0.3)
            
            # Overlay transaction volume
            if len(collapse_tx) > 0:
                daily_vol = collapse_tx.groupby('date')['value'].sum()
                daily_vol.index = pd.to_datetime(daily_vol.index)
                ax1_twin.bar(daily_vol.index, daily_vol.values / 1e6, 
                            alpha=0.3, color='orange', width=0.8, label='Daily Volume')
                ax1_twin.set_ylabel('Volume (Millions USD)', fontsize=11, color='orange')
                ax1_twin.tick_params(axis='y', labelcolor='orange')
                ax1_twin.legend(loc='upper right')
            
            # Plot 2: Loss distribution
            ax2 = axes[0, 1]
            loss_vis = estimated_losses[estimated_losses < 1000000]  # Filter for visualization
            ax2.hist(loss_vis, bins=50, edgecolor='black', alpha=0.7, color='darkred')
            ax2.set_xlabel('Estimated Loss (USD)', fontsize=11)
            ax2.set_ylabel('Number of Addresses', fontsize=11)
            ax2.set_title('Distribution of Estimated Losses', fontsize=12, fontweight='bold')
            ax2.grid(True, alpha=0.3)
            
            # Plot 3: Cumulative losses
            ax3 = axes[1, 0]
            cumulative_losses = estimated_losses.cumsum()
            ax3.plot(range(len(cumulative_losses)), cumulative_losses / 1e6, 
                    linewidth=2, color='red')
            ax3.fill_between(range(len(cumulative_losses)), 0, cumulative_losses / 1e6, 
                           alpha=0.3, color='red')
            ax3.set_xlabel('Address Rank (by Loss)', fontsize=11)
            ax3.set_ylabel('Cumulative Loss (Millions USD)', fontsize=11)
            ax3.set_title('Cumulative Loss Distribution', fontsize=12, fontweight='bold')
            ax3.grid(True, alpha=0.3)
            
            # Plot 4: Top loss bearers
            ax4 = axes[1, 1]
            top_losses = estimated_losses.head(15)
            ax4.barh(range(len(top_losses)), top_losses.values / 1e6, 
                    color='darkred', alpha=0.7)
            ax4.set_yticks(range(len(top_losses)))
            ax4.set_yticklabels([addr[:15] + '...' for addr in top_losses.index], 
                               fontsize=8)
            ax4.set_xlabel('Estimated Loss (Millions USD)', fontsize=11)
            ax4.set_title('Top 15 Estimated Loss Bearers', fontsize=12, fontweight='bold')
            ax4.grid(True, alpha=0.3, axis='x')
            
            plt.tight_layout()
            plt.savefig('q3_loss_estimation.png', dpi=300, bbox_inches='tight')
            plt.show()
            
            print(f"\n📊 Visualization saved as 'q3_loss_estimation.png'")
    else:
        print("⚠️ No price data available for collapse period")
else:
    print("⚠️ Price or transaction data not available")
```

```python
# ============================================================================
# QUESTION 3.3: SYSTEM DESIGN FAILURES - WHY PROTECTION FAILED
# ============================================================================

print("\n" + "=" * 80)
print("QUESTION 3.3: SYSTEM DESIGN FAILURES")
print("=" * 80)

# Analyze the algorithmic stablecoin mechanism failure
if 'USTC' in price_data and 'WLUNA' in price_data:
    # Get price data for both tokens
    ustc_price_analysis = price_data['USTC'][
        (price_data['USTC']['timestamp'] >= PRE_COLLAPSE_START) & 
        (price_data['USTC']['timestamp'] <= COLLAPSE_END)
    ].copy()
    
    wluna_price_analysis = price_data['WLUNA'][
        (price_data['WLUNA']['timestamp'] >= PRE_COLLAPSE_START) & 
        (price_data['WLUNA']['timestamp'] <= COLLAPSE_END)
    ].copy()
    
    # Merge on timestamp (hourly/daily basis)
    ustc_price_analysis['date'] = ustc_price_analysis['timestamp'].dt.date
    wluna_price_analysis['date'] = wluna_price_analysis['timestamp'].dt.date
    
    ustc_daily = ustc_price_analysis.groupby('date')['close'].mean().reset_index()
    wluna_daily = wluna_price_analysis.groupby('date')['close'].mean().reset_index()
    
    ustc_daily.columns = ['date', 'ustc_price']
    wluna_daily.columns = ['date', 'wluna_price']
    
    combined = ustc_daily.merge(wluna_daily, on='date', how='outer').sort_values('date')
    combined['date'] = pd.to_datetime(combined['date'])
    
    print("\n🔍 Algorithmic Stablecoin Mechanism Analysis:")
    print("   Terra's UST was an algorithmic stablecoin backed by LUNA")
    print("   When UST depegged, the arbitrage mechanism should have restored the peg")
    print("   However, the death spiral occurred instead:")
    print("   1. UST depegged below $1")
    print("   2. Arbitrageurs should buy UST and burn it for $1 worth of LUNA")
    print("   3. But LUNA price collapsed, breaking the mechanism")
    print("   4. This created a death spiral: more UST selling → more LUNA minting → LUNA price drops")
    
    # Calculate correlation
    correlation = combined[['ustc_price', 'wluna_price']].corr().iloc[0, 1]
    print(f"\n📊 Price Correlation (USTC vs WLUNA): {correlation:.4f}")
    print("   High positive correlation indicates both collapsed together")
    print("   This suggests the arbitrage mechanism failed")
    
    # Visualization
    fig, axes = plt.subplots(3, 1, figsize=(14, 12))
    
    # Plot 1: USTC and WLUNA prices
    ax1 = axes[0]
    ax1.plot(combined['date'], combined['ustc_price'], 
            linewidth=2, label='USTC Price', color='red', alpha=0.8)
    ax1_twin = ax1.twinx()
    ax1_twin.plot(combined['date'], combined['wluna_price'], 
                 linewidth=2, label='WLUNA Price', color='blue', alpha=0.8)
    ax1.axhline(y=1.0, color='green', linestyle='--', linewidth=2, alpha=0.5)
    ax1.set_xlabel('Date', fontsize=11)
    ax1.set_ylabel('USTC Price (USD)', fontsize=11, color='red')
    ax1_twin.set_ylabel('WLUNA Price (USD)', fontsize=11, color='blue')
    ax1.tick_params(axis='y', labelcolor='red')
    ax1_twin.tick_params(axis='y', labelcolor='blue')
    ax1.set_title('Death Spiral: USTC and WLUNA Price Collapse', fontsize=13, fontweight='bold')
    ax1.legend(loc='upper left')
    ax1_twin.legend(loc='upper right')
    ax1.grid(True, alpha=0.3)
    
    # Plot 2: Deviation from peg
    ax2 = axes[1]
    combined['ustc_deviation'] = (combined['ustc_price'] - 1.0) * 100
    ax2.plot(combined['date'], combined['ustc_deviation'], 
            linewidth=2, color='darkred', label='USTC Deviation from $1 Peg')
    ax2.axhline(y=0, color='green', linestyle='--', linewidth=1, alpha=0.5)
    ax2.fill_between(combined['date'], -5, 5, alpha=0.1, color='green', label='Stable Zone (±5%)')
    ax2.set_xlabel('Date', fontsize=11)
    ax2.set_ylabel('Deviation from Peg (%)', fontsize=11)
    ax2.set_title('USTC Peg Stability: Mechanism Failure', fontsize=13, fontweight='bold')
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    
    # Plot 3: Scatter plot showing relationship
    ax3 = axes[2]
    scatter_data = combined.dropna()
    if len(scatter_data) > 0:
        ax3.scatter(scatter_data['ustc_price'], scatter_data['wluna_price'], 
                   alpha=0.5, s=50, c=scatter_data['date'], cmap='viridis')
        ax3.set_xlabel('USTC Price (USD)', fontsize=11)
        ax3.set_ylabel('WLUNA Price (USD)', fontsize=11)
        ax3.set_title(f'USTC vs WLUNA: Correlation = {correlation:.3f}', 
                     fontsize=13, fontweight='bold')
        ax3.axvline(x=1.0, color='green', linestyle='--', alpha=0.5)
        ax3.grid(True, alpha=0.3)
        plt.colorbar(ax3.collections[0], ax=ax3, label='Date')
    
    plt.tight_layout()
    plt.savefig('q3_system_design_failure.png', dpi=300, bbox_inches='tight')
    plt.show()
    
    print(f"\n📊 Visualization saved as 'q3_system_design_failure.png'")
    
    # Key design failures
    print("\n❌ Key Design Failures:")
    print("   1. No Reserve Backing: UST relied solely on algorithmic mechanism")
    print("   2. Circular Dependency: UST stability depended on LUNA value, but LUNA value")
    print("      depended on UST demand")
    print("   3. No Circuit Breakers: No mechanism to halt trading during extreme volatility")
    print("   4. Unlimited Minting: Could mint unlimited LUNA, causing hyperinflation")
    print("   5. No Insurance Fund: No buffer to absorb losses during depegging")
else:
    print("⚠️ Price data not available for design analysis")
```
